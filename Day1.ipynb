```python
import numpy as np

x = np.array(range(10))
print(type(x), len(x), x)

print(x + x)
assert all(x + x == x * 2)
try:
    print(x - x)
except Exception as e:
    print("Error:", e)
```

```python
import pandas as pd
from sklearn.utils import Bunch

iris = Bunch()
print(type(iris))
iris['frame'] = pd.read_csv('iris.csv')
iris.feature_names = iris.frame.drop(columns='label').columns.tolist()
print(iris.feature_names)
iris.target_names = iris.frame['label'].unique()
print(iris.target_names)
iris.data = iris.frame.drop(columns='label').to_numpy()
print(iris.data.shape)
iris.frame.sample(5, random_state=1)
```

## 모형

모형(model)은 데이터 분포에 대한 가정을 반영해야 합니다.

```python
nums = np.arange(10)
pd.DataFrame({
    'nums < 5': nums < 5, 
    'np.where': np.where(nums < 5, '참', '거짓')}).T
```

```python
def 붓꽃분류(data):
    x3 = data.iloc[:, 2] if isinstance(data, pd.DataFrame) else data[:, 2]
    return np.where(x3 < 2.5, 0, np.where(x3 < 5.0, 1, 2))

예측 = 붓꽃분류(iris.data)
채점 = 예측 == iris.target
print(f'{채점.sum()}/{채점.size} = {채점.mean():.0%}')
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=1)
model.fit(iris.data, iris.target) # 지도 학습 (supervised learning)

예측 = model.predict(iris.data)
채점 = 예측 == iris.target
print(f'{채점.sum()}/{채점.size} = {채점.mean():.0%}')
```

## 훈련/시험

```python
시험데이터비율 = 0.25 # 25% (1/4)
시험데이터수 = round(len(iris.target) * 시험데이터비율)
print(시험데이터수)
# 훈련/시험 데이터 분리 (hold-out)
train_data = iris.data[:-시험데이터수]
train_target = iris.target[:-시험데이터수]
test_data = iris.data[-시험데이터수:]
test_target = iris.target[-시험데이터수:]
print(train_data.shape, test_data.shape)

# 데이터 분포 확인
print(pd.DataFrame({
    'all': iris.target.value_counts(),
    'train': train_target.value_counts(),
    'test': test_target.value_counts()
}))

# 모형 생성/학습
model = KNeighborsClassifier(n_neighbors=1)
model.fit(train_data, train_target) # 지도 학습 (supervised learning)
# 모형 평가
예측 = model.predict(test_data)
채점 = 예측 == test_target
print(f'{채점.sum()}/{채점.size} = {채점.mean():.0%}')
```
